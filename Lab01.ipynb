```python
# task 07 // extract domain from email
def extract_domain(email):
    return email.split('@')[1]
```

```python
print(extract_domain("user@domain.com"))
```

```python
# task 08 // function that gives True if dog is in the string
def find_dog(text):
    return "dog" in text.lower()   # text is the parameter that receives the string
```

```python
print(find_dog("is there a dog here?"))
```

```python
print(find_dog("Dog is friendly animal"))
```

```python
print(find_dog("i have a cat"))
```

```python
# task 09 // function that counts how many times dog appears in the string
def count_dog(text):
    return text.lower().count("dog")
```

```python
print(count_dog("dog dog DOG"))
```

```python
print(count_dog("the dog chased another dog"))
```

```python
print(count_dog("i have a cat"))
```

```python
# task 10 // caught_speeding(speed, is_birthday)
# on birthday we get 5 extra allowed speed
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed = speed - 5
    if speed <= 60:
        return "no ticket"
    elif speed <= 80:
        return "small ticket"
    else:
        return "big ticket"
```

```python
print(caught_speeding(60, False))
```

```python
print(caught_speeding(65, False))
```

```python
print(caught_speeding(65, True))
```

```python
print(caught_speeding(85, True))
```

```python
print(caught_speeding(90, False))
```

```python
print(caught_speeding(100, False))
```

```python
print(caught_speeding(40, False))
```
